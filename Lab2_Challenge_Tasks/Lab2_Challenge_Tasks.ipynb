```python
# [APPENDIX 1] 
ENGINEER = "Sebastian"
SEED_NUM = 7
limit = (SEED_NUM * 10) + 20  # Safety Limit: 90

# Generate a test set of raw sensor readings (includes valid numbers and invalid text)
raw_readings = [f"{SEED_NUM * 3}", "ERR_SENSOR", f"{limit + 15}", "2", f"{SEED_NUM * 10}"]

valid_readings = []
invalid_count = 0
classifications = {"Low Signal": 0, "Nominal": 0, "Over Limit": 0}

for entry in raw_readings:
    try:
        val = int(entry)
        valid_readings.append(val)
        if val > limit:
            classifications["Over Limit"] += 1
        elif val < 5:
            classifications["Low Signal"] += 1
        else:
            classifications["Nominal"] += 1
    except ValueError:
        invalid_count += 1

print("=== EXERCISE 1 ===")
print(f"Engineer: {ENGINEER} | Safety Limit: {limit}")
print(f"Raw Input Data: {raw_readings}")
print(f"Valid Readings: {valid_readings}")
print(f"Invalid Entries Handled: {invalid_count}")
print(f"Classification Breakdown: {classifications}")
```

```python
# [APPENDIX EXERCISE 2] 
ENGINEER = "Sebastian"
SEED_NUM = 7

# Generate unique signal string using surname and SEED_NUM
signal = f"{ENGINEER}_{SEED_NUM * 101}#SYS"

uppercase_cnt = 0
lowercase_cnt = 0
digits_cnt = 0
special_cnt = 0

print("=== EXERCISE 2===")
print(f"Target Signal: {signal}\n")

for char in signal:
    if char.isupper():
        uppercase_cnt += 1
        char_type = "Uppercase"
    elif char.islower():
        lowercase_cnt += 1
        char_type = "Lowercase"
    elif char.isdigit():
        digits_cnt += 1
        char_type = "Digit"
    else:
        special_cnt += 1
        char_type = "Special Character"
    print(f"Character: '{char}' -> {char_type}")

print("\n------")
print(f"Signal Length: {len(signal)} characters")
print(f"Uppercase: {uppercase_cnt} | Lowercase: {lowercase_cnt} | Digits: {digits_cnt} | Special: {special_cnt}")
```

```python
# [APPENDIX EXERCISE 3] 
ENGINEER = "Sebastian"
SEED_NUM = 7

# System credentials and limits derived from variables
password = f"{ENGINEER[:3].upper()}{SEED_NUM * 10}"  # Target: 'SEB70'
attempt_limit = SEED_NUM - 4  # 3 Max Attempts

print("=== EXERCISE 3 ===")
print(f"System Password: {password}")
print(f"Maximum Allowed Attempts: {attempt_limit}\n")

attempts_made = 0
access_granted = False

while attempts_made < attempt_limit:
    attempts_made += 1
    user_input = input(f"Attempt {attempts_made}/{attempt_limit} - Enter Password: ")
    
    if user_input == password:
        access_granted = True
        print("ACCESS GRANTED: System Unlocked successfully.")
        break
    else:
        remaining = attempt_limit - attempts_made
        print(f"ACCESS DENIED: Incorrect entry. ({remaining} attempts remaining)")

if not access_granted:
    print("\nSYSTEM LOCKED: Maximum authentication attempts reached!")
```
